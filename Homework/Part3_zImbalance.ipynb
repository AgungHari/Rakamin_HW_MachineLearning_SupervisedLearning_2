```python
#Cek imbalance pada y
import numpy as np
import pandas as pd
```

```python
y = pd.read_csv('y_train.csv', header=None).values.flatten()
```

```python
kelas, jumlah = np.unique(y, return_counts=True)

for k, j in zip(kelas, jumlah):
    print(f'Kelas {k}: {j} sampel')
```

```python
rasio = jumlah / jumlah.sum()

for k, r in zip(kelas, rasio):
    print(f'Kelas {k}: {r:.2%} dari total')
```
